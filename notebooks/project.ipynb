```python
import os
import sys

# Get the current folder where this notebook file is running (notebooks folder)
current_folder = os.getcwd()
print("Currently running inside folder:", current_folder)

# Go up one level to get to the main project folder
# os.path.dirname acts like clicking the "Up" arrow in file explorer
project_root_folder = os.path.dirname(current_folder)

# Add the project root directory to sys.path so Python looks there for modules
if project_root_folder not in sys.path:
    sys.path.append(project_root_folder)
    print("Successfully added parent directory to path!")
else:
    print("Parent directory was already added to path.")

print("Python will now check this path for modules:", project_root_folder)
```

```python
# Due to path difficulties, the import function is being tested
try:
    # This looks for log_parser.py and grabs the read_logs function
    from log_parser import read_logs
    print("SUCCESS: 'read_logs' function imported perfectly from log_parser.py!")
except ModuleNotFoundError:
    print("ERROR: Still having issues")
```

```python
# Step 1: Define the file path to our sample logs
log_file_path = "../logs/sample_log.txt"

# Step 2: Use the working function to load the log entries into a list
logs = read_logs(log_file_path)

print("--- PARSED LOG OUTPUT (CLEAN DISPLAY) ---")

# Step 3: Loop through the list line-by-line to stack them vertically
for line in logs:
    # .strip() removes the ugly hidden '\n' characters so it formats normally
    print(line.strip())
```

```python
# Step 4: Test the alert functions to ensure they are properly imported and ready to use
from alert_generator import detect_bruteforce
from alert_generator import detect_sql
from alert_generator import detect_xss

print("Alert functions imported successfully.")
```

```python
# Step 5: Run the alert functions on the loaded logs and print out the results
bruteforce_count = detect_bruteforce(logs)
sql_count = detect_sql(logs)
xss_count = detect_xss(logs)

print("====== ALERT SUMMARY ======")
print("Failed Logins:", bruteforce_count)
print("SQL Injection Attempts:", sql_count)
print("XSS Attempts:", xss_count)
```

```python
# Step 6: Print out warnings if any suspicious activity is detected
print("Analysis Complete")

if bruteforce_count > 0:
    print("WARNING: FAILED LOGIN ATTEMPTS DETECTED.")

if sql_count > 0:
    print("WARNING: POSSIBLE SQL INJECTION ACTIVITY DETECTED.")

if xss_count > 0:
    print("WARNING: POSSIBLE XSS ACTIVITY DETECTED.")
```

```python
# Step 7: Generate a simple report summarizing the findings and save it to the security report
report = f"""
ALERT SUMMARY REPORT

========================

Failed Login Attempts: {bruteforce_count}
SQL Injection Attempts: {sql_count}
XSS Attempts: {xss_count}

========================
"""

with open("../reports/security_report.txt", "w") as file:
    file.write(report)

print("Report Successfully Generated.")
```
